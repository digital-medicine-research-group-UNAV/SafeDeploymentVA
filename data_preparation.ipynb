```python
import os
import sys

import numpy as np
import pandas as pd
from ucimlrepo import fetch_ucirepo 
from sklearn.preprocessing import LabelEncoder
```

```python
# german
german = fetch_ucirepo(id=144) 
  
X = german.data.features 
y = german.data.targets

# Encode using label encoding
le = LabelEncoder()
for col in X.select_dtypes(include=['object']).columns:
    X[col] = le.fit_transform(X[col])
    
# Change the name of the target column
y.columns = ["Class"]
y["Class"] = y["Class"] - 1

# Merge both into a single dataframe
data = pd.concat([X, y], axis=1)

data.to_csv("data/german.csv", index=False)
```
